```python
import os
import warnings
import itertools
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


from matplotlib.colors import LinearSegmentedColormap
from typing import Optional
from hydra.utils import to_absolute_path
from mlflow import set_tracking_uri, get_experiment_by_name, search_runs
from scipy.stats import percentileofscore, kendalltau, spearmanr
from typing import Iterable, List, Tuple, Optional, Set


# TODO: Adjust global constants as needed.
MLRUNS_PATH = "/mnt/work/mherde/maml/crowd_hpo/"
FIGURES_PATH = "/mnt/home/mherde/projects/github/multi-annotator-machine-learning/figures/"
CACHE_PATH = "/mnt/home/mherde/projects/github/multi-annotator-machine-learning/empirical_evaluation/python_scripts/results_hpo_backup"

SHOW_LATEX = False
SHOW_INDIVIDUAL_SCORES = False
DATASET_GROUP = "all" # low, high, all
CANDIDATE_SIZE = None # None, 31, 11

# Filter meaningless warnings.
warnings.filterwarnings("ignore")

# Define global color setings.
plt.rcParams["font.family"] = "serif"
plt.rcParams["mathtext.fontset"] = "dejavuserif"
colors = [
    (0.0, (0, 128/255, 128/255)),   
    (0.5, (1.0, 1.0, 1.0)),           
    (1.0, (127/255, 0, 127/255)), 
]
custom_cmap = LinearSegmentedColormap.from_list("DivergingTealPurple", colors)

# Increase maximum number of displayed rows.
pd.set_option("display.max_rows", 1000)

# Define global constants for the visualization of results.
UPDATE_COLUMNS = {
    "params.data.optimizer.gt_params.lr": "gt_lr",
    "params.data.optimizer.gt_params.weight_decay": "gt_wd",
    "params.data.optimizer.ap_params.lr": "ap_lr",
    "params.data.optimizer.ap_params.weight_decay": "ap_wd",
    "params.data.train_batch_size": "bs",
    "params.data.max_epochs": "me",
    "params.data.lr_scheduler.params.T_max": "T_max",
    "params.architecture.params.dropout_rate": "dr",
    "params.classifier.params.lmbda": "lmbda",
    "params.classifier.params.eta": "eta",
    "params.classifier.params.alpha": "alpha",
    "params.classifier.params.beta": "beta",
    "params.classifier.embed_size": "dim",
    "params.classifier.params.epsilon": "epsilon",
    "params.classifier.params.mu": "mu",
    "params.architecture.params.n_hidden_neurons.0": "h_dim_0",
    "params.architecture.params.n_hidden_neurons.1": "h_dim_1",
}
INV_UPDATE_COLUMNS = {v: k for k, v in UPDATE_COLUMNS.items()}
APPROACHES = [
    "ground-truth",
    "majority-vote",
    "dawid-skene",
    "crowd-layer",
    "trace-reg",
    "conal",
    "union-net-a",
    "union-net-b",
    "geo-reg-w",
    "geo-reg-f",
    "madl",
    "crowd-ar",
    "annot-mix",
    "coin-net",
]
    
agg_and_classcorr_noise = {

    
    # label-me (C=8)
    "label_me_worst-1": (41.1, 47.0, 8),
    "label_me_worst-2": (30.1, 34.4, 8),
    "label_me_worst-var": (32.5, 37.1, 8),
    "label_me_rand-1":  (23.9, 27.3, 8),
    "label_me_rand-2":  (25.7, 29.4, 8),
    "label_me_rand-var":  (25.0, 28.6, 8),
    "label_me_full":    (23.7, 27.1, 8),
    
    # dopanim (C=15)
    "dopanim_worst-1": (77.6, 83.1, 15),
    "dopanim_worst-2": (62.2, 66.6, 15),
    "dopanim_worst-var": (46.9, 50.2, 15),
    "dopanim_rand-1":  (32.5, 34.8, 15),
    "dopanim_rand-2":  (33.2, 35.6, 15),
    "dopanim_rand-var":  (26.3, 28.2, 15),
    "dopanim_full":    (19.3, 20.7, 15),
    
    # mgc (C=10)
    "music_genres_worst-1": (87.4, 97.1, 10),
    "music_genres_worst-2": (69.4, 77.1, 10),
    "music_genres_worst-var": (58.6, 65.1, 10),
    "music_genres_rand-1":  (47.1, 52.3, 10),
    "music_genres_rand-2":  (43.9, 48.8, 10),
    "music_genres_rand-var":  (38.3, 42.6, 10),
    "music_genres_full":    (30.3, 33.7, 10),
    
    
    # reuters (C=8)
    "reuters_worst-1": (69.2, 79.1, 8),
    "reuters_worst-2": (54.0, 61.7, 8),
    "reuters_worst-var": (51.8, 59.2, 8),
    "reuters_rand-1":  (38.5, 44.0, 8),
    "reuters_rand-2":  (40.9, 46.7, 8),
    "reuters_rand-var":  (38.4, 43.9, 8),
    "reuters_full":    (35.5, 40.6, 8),
    
    # spc (C=2) — values >100% mean below-chance accuracy
    "spc_worst-1": (63.4, 126.8, 2),
    "spc_worst-2": (47.0, 94.0, 2),
    "spc_worst-var": (32.5, 65.0, 2),
    "spc_rand-1":  (21.2, 42.4, 2),
    "spc_rand-2":  (20.6, 41.2, 2),
    "spc_rand-var":  (14.9, 29.8, 2),
    "spc_full":    (11.0, 22.0, 2),
}

DATASETS = []
noise_adj_list = []
for d, (noise_orig, noise_adj, _) in agg_and_classcorr_noise.items():
    if DATASET_GROUP == "low" and noise_adj <= 43.9:
        DATASETS.append(d)
    elif DATASET_GROUP == "high" and noise_adj > 43.9:
        DATASETS.append(d)
    elif DATASET_GROUP == "all":
        DATASETS.append(d)
print(len(DATASETS))
    
#DATASETS = []
#for v in ["worst-1", "worst-2", "worst-var", "rand-1", "rand-2", "rand-var", "full"]:
#    DATASETS.extend(
#        [
#        f"music_genres_{v}",
#        f"label_me_{v}",
#        f"dopanim_{v}",
#        f"reuters_{v}",
#        f"spc_{v}",
#        ]
#    )
    
LOSS_FUNC = "zero_one_loss" # brier_score, zero_one_loss
LOSS_FUNC_CRITERIA = "zero_one_loss"

def process_df(df, clf_col='clf', agg_col='agg'):
    """
    Process a DataFrame by combining classifier and aggregator columns, cleaning specific substrings, 
    dropping the aggregator column, and reordering columns.

    Parameters
    ----------
    df : pandas.DataFrame
        DataFrame containing at least the columns specified by `clf_col` and `agg_col`.
    clf_col : str, optional
        The name of the classifier column. Default is 'clf'.
    agg_col : str, optional
        The name of the aggregator column. Default is 'agg'.

    Returns
    -------
    pandas.DataFrame
        The processed DataFrame with a combined column (named after `clf_col`) placed at the beginning, 
        with the original `agg_col` column removed and specific substrings ('aggregate_' and '_None') replaced 
        with empty strings.
    """
    # Reorder columns so that the combined column comes first
    cols = [clf_col] + [col for col in df.columns if col != clf_col]
    return df[cols]

def same_pairs_per_group_farthest_first(
    df: pd.DataFrame,
    group_cols: Iterable[str],
    pair_cols: Optional[Iterable[str]] = None,
    n: Optional[int] = None,
    frac: Optional[float] = None,
    force_keep: Optional[Iterable[Tuple]] = None,
):
    """
    Per group (defined by `group_cols`), keep k rows chosen by farthest-first over
    'params.*' columns. For each group, drop any 'params.*' column that is not fully
    numeric *within that group*. Rows are never dropped due to non-numeric params.

    - k is uniform across groups:
        * if n provided: k = n
        * elif frac provided: k = min(floor(frac * |group_i|))
        * else: k = min group size
    - force_keep: list of tuples matching `pair_cols` order. In each group, the first
      occurrence of each forced tuple is reserved toward k.

    Returns
    -------
    out_df : pd.DataFrame
    k_used : int
    """
    # sanitize
    if isinstance(group_cols, str): group_cols = [group_cols]
    group_cols = list(group_cols)
    if pair_cols is not None and isinstance(pair_cols, str): pair_cols = [pair_cols]
    pair_cols = list(pair_cols) if pair_cols is not None else None

    # checks
    missing = [c for c in group_cols if c not in df.columns]
    if missing: raise KeyError(f"Group columns not in df: {missing}")
    if pair_cols:
        pmiss = [c for c in pair_cols if c not in df.columns]
        if pmiss: raise KeyError(f"pair_cols not in df: {pmiss}")

    params_cols_all = ['gt_lr', 'gt_wd','bs', 'me', 'dr', 'lmbda', 'eta', 'alpha', 'beta', 'dim', 'epsilon', "mu"]
    params_scale_details = {
        "all": {'gt_lr': ("log", 1e-4, 1e-1), 'gt_wd': ("log", 1e-6, 1e-3), 'bs': ("lin", 16, 64), 'me': ("lin", 5, 50), 'dr': ("lin", 0, 0.5)},
        "trace_reg": {"lmbda": ("log", 1e-3, 1e-1)},
        "conal": {"lmbda": ("log", 1e-6, 1e-3), "dim": ("lin", 20, 80)},
        "union_net_a": {"epsilon": ("log", 1e-6, 1e-4)},
        "union_net_b": {"epsilon": ("log", 1e-6, 1e-4)},
        "geo_reg_w": {"lmbda": ("log", 1e-4, 1e-2)},
        "geo_reg_f": {"lmbda": ("log", 1e-4, 1e-2)},
        "madl": {"eta": ("lin", 0.75, 0.95), "alpha": ("lin", 1.0, 1.5), "beta": ("lin", 0.25, 0.5), "dim": ("lin", 8, 32)},
        "crowd_ar": {"mu": ("lin", 0.0, 1.0)},
        "annot_mix": {"eta": ("lin", 0.75, 0.95), "alpha": ("lin", 0.0, 2.0)},
        "coin_net": {"lmbda": ("log", 1e-3, 1e-1), "mu": ("lin", 0.0, 1.0)},
    }
        
            
    # stable base order for deterministic tie-breaking
    dfr = df.reset_index(drop=False).rename(columns={"index": "_orig_idx"})
    base_sort = group_cols + (pair_cols or []) + ["_orig_idx"]
    dfr = dfr.sort_values(base_sort, kind="mergesort")

    # group sizes (no dropping of rows)
    gsize = dfr.groupby(group_cols, sort=False).size()
    if gsize.empty:
        return df.iloc[0:0].copy(), 0

    # decide k
    if n is not None and frac is not None:
        raise ValueError("Provide only one of n or frac.")
    if n is not None:
        k = int(n)
    elif frac is not None:
        k = int(np.floor((gsize.values * float(frac)).min()))
    else:
        k = int(gsize.min())

    if k < 0:
        raise ValueError("k must be non-negative.")
    if k > gsize.min():
        raise ValueError(f"Requested k={k} exceeds smallest group size={int(gsize.min())}.")

    # force_keep
    force_set = set(map(tuple, force_keep)) if force_keep is not None else set()
    if force_set and not pair_cols:
        raise ValueError("force_keep requires pair_cols to match tuples.")

    keep_idx = []

    for grp_vals, grp_df in dfr.groupby(group_cols, sort=False):
        # Determine usable params columns for THIS group:
        # a params column is usable if all its values in this group can be coerced to numeric
        usable_cols = []
        for c in params_cols_all:
            s = pd.to_numeric(grp_df[c], errors="coerce")
            if not s.isna().any():
                usable_cols.append(c)
                

                
        def to_unit_log(x, lo, hi):
            x = np.asarray(x, float)
            return (np.log(x) - np.log(lo)) / (np.log(hi) - np.log(lo))

        def to_unit_lin(x, lo, hi):
            x = np.asarray(x, float)
            return (x - lo) / (hi - lo)

        # Build feature matrix; if none usable, use a single zero feature (degenerate distances)
        if usable_cols:
            X = grp_df[usable_cols].apply(pd.to_numeric, errors="coerce").to_numpy(dtype=float)
            Xz = X.copy()
            for u_idx, u_col in enumerate(usable_cols):
                z = Xz[:, u_idx]
                if u_col in params_scale_details["all"].keys():
                    u_scale, u_lo, u_hi = params_scale_details["all"][u_col]
                elif grp_vals[1] in params_scale_details.keys() and u_col in params_scale_details[grp_vals[1]].keys():
                    u_scale, u_lo, u_hi = params_scale_details[grp_vals[1]][u_col]
                else:
                    u_scale = None
                if u_scale == "log":
                    z = to_unit_log(z, u_lo, u_hi)
                elif u_scale == "lin":
                    z = to_unit_lin(z, u_lo, u_hi)
                Xz[:, u_idx] = z
        else:
            # no numeric params in this group; distances collapse, selection falls back to order
            Xz = np.zeros((len(grp_df), 1), dtype=float)

        # Resolve forced rows: first occurrence of each tuple present in this group
        reserved = []
        if force_set:
            tuples_grp = list(map(tuple, grp_df[pair_cols].to_numpy()))
            seen = set()
            for pos, t in enumerate(tuples_grp):
                if t in force_set and t not in seen:
                    reserved.append(pos)
                    seen.add(t)
                    if len(reserved) == k:  # early stop
                        break
        if len(reserved) > k:
            raise ValueError(
                f"Group {grp_vals}: force_keep needs {len(reserved)} > k={k}."
            )

        # Farthest-first selection
        selected = list(reserved)
        need = k - len(selected)
        if need > 0:
            N = Xz.shape[0]
            all_pos = np.arange(N)
            cand_mask = np.ones(N, dtype=bool)
            if selected:
                cand_mask[np.array(selected, dtype=int)] = False
            cand = all_pos[cand_mask]

            if len(selected) == 0:
                # seed: farthest from centroid; ties break by order
                c = Xz[cand].mean(axis=0)                                  # centroid of candidates
                norms = np.sum((Xz[cand] - c) ** 2, axis=1)                # squared distance to centroid
                seed_rel = int(np.argmax(norms))
                seed = int(cand[seed_rel])
                selected.append(seed)
                cand = cand[cand != seed]
                need -= 1

            # initialize min distances to current selected set
            if need > 0:
                min_d = np.full(len(cand), np.inf)
                for s in selected:
                    d = np.sum((Xz[cand] - Xz[s]) ** 2, axis=1)
                    min_d = np.minimum(min_d, d)

                while need > 0 and len(cand) > 0:
                    pick_rel = int(np.argmax(min_d))
                    pick = int(cand[pick_rel])
                    selected.append(pick)
                    cand = np.delete(cand, pick_rel)
                    if len(cand) == 0:
                        break
                    d_new = np.sum((Xz[cand] - Xz[pick]) ** 2, axis=1)
                    min_d = np.minimum(np.delete(min_d, pick_rel), d_new)
                    need -= 1

        # Map back to original dataframe indices
        keep_idx.extend(grp_df.index.to_numpy()[np.array(selected, dtype=int)])

    out = (
        dfr.loc[keep_idx]
           .sort_values("_orig_idx", kind="mergesort")
           .drop(columns=["_orig_idx"])
           .reset_index(drop=True)
    )
    return out, k

def evaluate(
    mlruns_path: str,
    experiment_name: str,
    update_columns: dict = None,
    perf_type: str = "gt",
    version: str = "test",
    metric: str = "acc",
    epoch: str = "best",
    loss_func: str = "zero_one_loss",
    print_number_of_runs: bool = False,
    cache_path: Optional[str] = None,
):
    """
    Queries the evaluation results via mlflow.

    Parameters
    ----------
    mlruns_path : str
        Path to the results saved via mlflow.
    experiment_name : str
        Name of the mlflow experiment.
    update_columns : dict, default=None
        Optional dictionary of columns to be included in the output table.
        If None, the global UPDATE_COLUMNS is used.
    perf_type : str, default="gt"
        'gt' for ground-truth estimates or 'ap' for annotator performance estimates.
    version : str, default="test"
        One of 'train', 'valid', or 'test' for different data subsets.
    metric : str, default="acc"
        Performance metric: 'acc', 'brier_score', or 'tce'.
    epoch : str, default="best"
        Either 'last' or 'best' epoch.
    print_number_of_runs : bool, default=False
        Whether to print the count of runs per classifier.
    cache_path : Optional[str], default=None
        If provided, results are cached/read from the given path.

    Returns
    -------
    runs: pd.DataFrame or None
        Table of results or None if no experiment/runs found.
    """
    runs = None
    
    # Use the provided update_columns or fall back to the default global constant.
    if update_columns is None:
        update_columns = UPDATE_COLUMNS.copy()
    
    # Attempt to load from cache if available.
    if cache_path:
        cache_file = os.path.join(cache_path, f"{experiment_name}.csv")
        if os.path.isfile(cache_file):
            runs = pd.read_csv(cache_file)
            if "hyperparameter_search" in experiment_name:
                runs = runs.append(pd.read_csv(cache_file.replace("hyperparameter_search", "default")))
    
    if runs is None:
        # Set mlflow tracking URI.
        abs_mlruns_path = to_absolute_path(mlruns_path)
        set_tracking_uri(uri=f"file://{abs_mlruns_path}")

        exp = get_experiment_by_name(experiment_name)
        if exp is None:
            return None

        query = "status = 'FINISHED'"
    
        runs = search_runs(experiment_ids=exp.experiment_id, filter_string=query, output_format="pandas")
        if runs.empty:
            return None
        if cache_path:
            try:
                runs.to_csv(cache_file, index=False)
            except Exception as e:
                print(f"Warning: Unable to cache results due to error: {e}")
                
    # Base column renaming.
    rename_columns = {
        "params.data.class_definition._target_": "data",
        "params.classifier.name": "clf",
        "params.classifier.aggregation_method": "agg",
    }
    # Update renaming mapping if provided columns exist in runs.
    for col, new_name in update_columns.items():
        if col in runs.columns:
            rename_columns[col] = new_name

    # Build aggregation dictionary.
    agg_dict = {
        f"metrics.{perf_type}_true_{loss_func}_{epoch}_test": ["mean", "std"],
        "params.seed": ["sum"],
    }
    bayes_columns = []
    for col in runs.columns:
        if col.startswith("metrics.") and col.endswith("_valid") and "bayes" not in col:
            agg_dict[col] = ["mean", "std"]

    # Drop duplicates using selected columns.
    # Note: Ensure that 'params.data.class_definition.realistic_split' exists in runs.
    dup_keys = list(rename_columns.keys()) + ["params.seed", "params.data.class_definition.realistic_split"]
    runs = runs.drop_duplicates(subset=dup_keys)
    
    if print_number_of_runs:
        # Actually print the value counts.
        print(runs["params.classifier.name"].value_counts())

    runs = runs.sort_values(by="params.seed")
    runs = runs.groupby(list(rename_columns.keys()), as_index=False, dropna=False).agg(agg_dict, skipna=False)

    # Flatten multi-index columns if necessary.
    if isinstance(runs.columns, pd.MultiIndex):
        runs.columns = ["".join(map(str, col)).strip() for col in runs.columns.values]

    # Round 'metrics' columns for accuracy if they exist.
    for col in runs.columns:
        if col.startswith("metrics") and "zero_one_loss" in col:
            runs[col] = np.round(runs[col] * 100, 3)

    # Add an extra column mapping.
    rename_columns["params.seedsum"] = "n_runs"

    # Build a combined renaming mapping for any remaining columns.
    additional_rename = {
        col: col.replace("mean", "")
        .replace("max", "")
        .replace("std", "_std")
        .replace("-", "_")
        .replace("metrics.", "")
        for col in runs.columns
        if col not in rename_columns
    }
    # Merge both dictionaries. (Note: The "|" operator requires Python 3.9+)
    full_rename = rename_columns | additional_rename
    runs = runs.rename(columns=full_rename)
    
    if "hyperparameter_search" in experiment_name and CANDIDATE_SIZE is not None:
        runs, _ = same_pairs_per_group_farthest_first(
            runs,
            group_cols=["data", "clf", "agg"],
            pair_cols=["gt_lr", "gt_wd", "bs", "me"],
            n=CANDIDATE_SIZE,
            force_keep=[(0.001, 0.0, 32, 30)]
        )
        print(len(runs))
    
    return runs
```

```python
# --- Add after your current imports ---
import itertools
import numpy as np
import pandas as pd
from scipy.stats import friedmanchisquare, wilcoxon
from statsmodels.stats.multitest import multipletests  # pip install statsmodels

alpha = 0.05
lower_is_better = True  # your W/L logic uses "<"
equalize_datasets = False  # True → give each dataset equal weight by averaging its sets

# ---------------- Build blocks: one row = one independent set ----------------
blocks = []
counts_per_ds = {}

for ds_name, df in df_dict.items():
    # same selection you used for W/L
    dfm = df["mean"][ordered_criteria][1:].copy()   # rows = independent sets
    if dfm.empty:
        continue
    dfm["__ds__"] = ds_name
    blocks.append(dfm)
    counts_per_ds[ds_name] = len(dfm)

if not blocks:
    raise RuntimeError("No data found after selection.")

X_all = pd.concat(blocks, axis=0, ignore_index=False)

if equalize_datasets:
    # Average across sets per dataset so each dataset contributes once
    X_for_tests = (
        X_all.drop(columns="__ds__", errors="ignore")
             .assign(__ds__=X_all["__ds__"].values)
             .groupby("__ds__", as_index=True).mean(numeric_only=True)
    )
else:
    # Use every set as its own block
    X_for_tests = X_all.drop(columns="__ds__", errors="ignore")

# Drop methods with all-NaN and require complete pairing
X_for_tests = X_for_tests.dropna(axis=1, how="all")
Xc = X_for_tests.dropna(axis=0, how="any")
methods = list(Xc.columns)

N_blocks, K = Xc.shape
print(f"[Info] Friedman input: {N_blocks} blocks × {K} methods "
      f"({('equalized per dataset' if equalize_datasets else 'per set as block')}).")
print("[Info] Blocks per dataset:", {k: int(v) for k, v in counts_per_ds.items()})

if N_blocks < 2 or K < 2:
    raise RuntimeError("Not enough data for a meaningful Friedman test.")

# ---------------- Friedman omnibus ----------------
fried_stat, fried_p = friedmanchisquare(*[Xc[m].values for m in methods])
print(f"[Friedman] chi2={fried_stat:.3f}, p={fried_p:.4g}")

# ---------------- Pairwise post-hoc if significant ----------------
sig_win = pd.DataFrame(False, index=methods, columns=methods, dtype=bool)  # i significantly better than j?

if fried_p < alpha:
    records, pvals = [], []
    for a, b in itertools.combinations(methods, 2):
        xa, xb = Xc[a].to_numpy(), Xc[b].to_numpy()
        try:
            res = wilcoxon(xa, xb, alternative="two-sided", zero_method="wilcox", method="auto")
            p = res.pvalue
        except ValueError:
            p = 1.0
        diff = float(np.mean(xa - xb))  # negative ⇒ a<b on average (better if lower_is_better)
        records.append({"a": a, "b": b, "p": p, "mean_diff": diff})
        pvals.append(p)

    reject, p_adj, _, _ = multipletests(pvals, alpha=alpha, method="holm")
    for rec, pa, rj in zip(records, p_adj, reject):
        if not rj:
            continue
        if lower_is_better:
            if rec["mean_diff"] < 0:  # a better than b
                sig_win.loc[rec["a"], rec["b"]] = True
            elif rec["mean_diff"] > 0:  # b better than a
                sig_win.loc[rec["b"], rec["a"]] = True
        else:
            if rec["mean_diff"] > 0:
                sig_win.loc[rec["a"], rec["b"]] = True
            elif rec["mean_diff"] < 0:
                sig_win.loc[rec["b"], rec["a"]] = True
else:
    print("[Post-hoc] Omnibus not significant; no pairwise stars will be shown.")

# ---------------- Align to your W/L display order and overlay stars ----------------
sig_win = sig_win.reindex(index=winning_percentage.index, columns=winning_percentage.columns)

fig, ax = plt.subplots(figsize=(15, 5.25))
ax.pcolormesh(
    winning_diffs, cmap=custom_cmap, vmin=0, vmax=1,
    alpha=0.3, edgecolors="white", linewidth=2, shading="flat"
)
ax.set_aspect("auto")

for i in range(winning_percentage.shape[0]):
    for j in range(winning_percentage.shape[1]):
        val = winning_percentage.iloc[i, j]
        txt = "N/A" if pd.isna(val) else f"{val:.2f}"
        if (not pd.isna(val)) and bool(sig_win.iloc[i, j]):
            star = " +" 
        elif (not pd.isna(val)) and bool(sig_win.iloc[j, i]):
            star = " -" 
        else:
            star = ""
        plt.text(j+0.5, i+0.56, rf'${txt}${star}', ha="center", va="center", color="black", fontsize=18)

ax.set_xticks(np.arange(len(winning_percentage))+0.5, winning_percentage.index.values, fontsize=12)
ax.set_yticks(np.arange(len(winning_percentage))+0.5, winning_percentage.index.values, fontsize=12)
ax.invert_yaxis()
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_PATH, f"criteria_winning_matrix_{LOSS_FUNC}.pdf"))
plt.show()
```

```python
# --- extra imports ---
import itertools
from scipy.stats import friedmanchisquare, wilcoxon
from statsmodels.stats.multitest import multipletests  # pip install statsmodels

alpha = 0.05
lower_is_better = True  # your win logic uses "<"

# Initialize DataFrames to hold win counts and total comparisons for each pair (i,j)
win_counts = None
total_counts = None

# We'll also collect per-dataset vectors for significance tests
_test_rows = []
_test_ds = []

# Loop over each dataset and each variant (row)
for ds_name, df in df_dict.items():
    dfm = df["mean"][["ENS"]][1:]  # rows = variants, single column ENS

    if dfm.empty:
        continue

    # init W/L matrices on first dataset
    if win_counts is None or total_counts is None:
        win_counts = pd.DataFrame(0, index=dfm.index, columns=dfm.index, dtype=float)
        total_counts = pd.DataFrame(0, index=dfm.index, columns=dfm.index, dtype=float)

    # W/L counting exactly as you had it
    for i, row_i in dfm.iterrows():
        for j, row_j in dfm.iterrows():
            if pd.notnull(row_i.values[0]) and pd.notnull(row_j.values[0]):
                total_counts.loc[i, j] += 1
                if row_i.values[0] < row_j.values[0]:
                    win_counts.loc[i, j] += 1

    # collect this dataset as one "block" for tests
    s = dfm["ENS"].copy()
    s.name = ds_name
    _test_rows.append(s)
    _test_ds.append(ds_name)

# Compute winning percentage matrix (expressed as a percentage)
winning_percentage = win_counts / total_counts * 100
winning_diffs = ((winning_percentage.values.T - winning_percentage.values) > 0).astype(float)
np.fill_diagonal(winning_diffs, np.nan)
np.fill_diagonal(winning_percentage.values, np.nan)

# ======================= significance overlay ===========================
# Build wide table: rows = datasets (blocks), cols = variants (methods)
X = pd.DataFrame(_test_rows, index=_test_ds)  # union of variant indices across datasets
# Friedman requires complete blocks; pairwise Wilcoxon can use pairwise-complete data
X_friedman = X.dropna(axis=1, how="all").dropna(axis=0, how="any")

methods = list(X_friedman.columns)
N_blocks, K = X_friedman.shape
print(f"[Info] Friedman input: {N_blocks} blocks × {K} variants")

sig_win = pd.DataFrame(False, index=X.columns, columns=X.columns, dtype=bool)  # i significantly better than j?

fried_p = 1.0
if N_blocks >= 2 and K >= 2:
    fried_stat, fried_p = friedmanchisquare(*[X_friedman[m].values for m in methods])
    print(f"[Friedman] chi2={fried_stat:.3f}, p={fried_p:.4g}")
else:
    print("[Friedman] Not enough complete blocks; skipping omnibus and stars.")

if fried_p < alpha:
    # Pairwise Wilcoxon with Holm correction (two-sided), using all available blocks per pair
    records, pvals = [], []
    for a, b in itertools.combinations(X.columns, 2):
        xa, xb = X[a], X[b]
        mask = xa.notna() & xb.notna()
        xa, xb = xa[mask].to_numpy(), xb[mask].to_numpy()
        if xa.size < 2:
            p = 1.0
            mean_diff = np.nan
        else:
            try:
                res = wilcoxon(xa, xb, alternative="two-sided", zero_method="wilcox", method="auto")
                p = res.pvalue
            except ValueError:
                p = 1.0
            mean_diff = float(np.mean(xa - xb))  # negative ⇒ a<b on average (better if lower_is_better)
        records.append({"a": a, "b": b, "p": p, "mean_diff": mean_diff})
        pvals.append(p)

    reject, p_adj, _, _ = multipletests(pvals, alpha=alpha, method="holm")
    for rec, pa, rj in zip(records, p_adj, reject):
        if not rj or np.isnan(rec["mean_diff"]):
            continue
        if lower_is_better:
            if rec["mean_diff"] < 0:  # a better than b
                sig_win.loc[rec["a"], rec["b"]] = True
            elif rec["mean_diff"] > 0:  # b better than a
                sig_win.loc[rec["b"], rec["a"]] = True
        else:
            if rec["mean_diff"] > 0:
                sig_win.loc[rec["a"], rec["b"]] = True
            elif rec["mean_diff"] < 0:
                sig_win.loc[rec["b"], rec["a"]] = True
else:
    print("[Post-hoc] Omnibus not significant; no pairwise stars will be shown.")

# Align significance matrix to your display order
sig_win = sig_win.reindex(index=winning_percentage.index, columns=winning_percentage.columns)

# ======================= plot with stars ================================
fig, ax = plt.subplots(figsize=(15, 6))
ax.pcolormesh(
    winning_diffs, cmap=custom_cmap, vmin=0, vmax=1,
    alpha=0.3, edgecolors="white", linewidth=2, shading="flat"
)
ax.set_aspect("auto")
for i in range(winning_percentage.shape[0]):
    for j in range(winning_percentage.shape[1]):
        formatted_text = r'${:.2f}$'.format(winning_percentage.iloc[i, j]).replace("$nan$", "N/A")
        star = " +" if (formatted_text != "N/A") and bool(sig_win.iloc[i, j]) else ""
        if (formatted_text != "N/A") and bool(sig_win.iloc[i, j]):
            star = " +" 
        elif (formatted_text != "N/A") and bool(sig_win.iloc[j, i]):
            star = " -" 
        else:
            star = ""
        plt.text(j+0.5, i+0.55, formatted_text + star, ha="center", va="center", color="black", fontsize=18)

ax.set_xticks(np.arange(len(winning_percentage))+0.5, winning_percentage.index.values, fontsize=12)
ax.set_yticks(np.arange(len(winning_percentage))+0.5, winning_percentage.index.values, fontsize=12)
ax.invert_yaxis()
plt.tight_layout()
plt.savefig(os.path.join(FIGURES_PATH, f"approaches_winning_matrix_{LOSS_FUNC}.pdf"))
plt.show()
```

```python
def fmt_number(x: float) -> str:
    s = f"{x:.2f}"
    if x < 10:
        s = r"\phantom{1}" + s
    return s

def fmt_mean(name: str, x: float, add_plus=True) -> str:
    s = fmt_number(x)
    if name == best:
        s = r"\textBF{" + s + "}"
    elif name == second:
        s = r"\underline{" + s + "}"
    if add_plus:
        return f"$+{s}$"
    else:
        return "$\phantom{+}" + s + "$" 

def fmt_se(x: float) -> str:
    return r"$\pm" + fmt_number(x) + r"$"
```

# Zero–one Loss Reductions of HPS Criteria per LFC Approach

For each LFC approach ($x$-axis), the scatter plot displays the mean and standard error of a criterion's reduction in zero–one loss ($y$-axis) compared to the criterion DEF.

```python
import pandas as pd
import numpy as np

for scatter_type in ["absolute", "relative"]:
    # ------------------------------------------------------------------
    # SETTINGS
    # ------------------------------------------------------------------
    approaches      = list(subsets.keys())
    approaches.insert(1, "DEF-DATA")
    approaches.insert(2, "DEF")
    baseline        = "DEF"
    comp_approaches = [a for a in approaches if a != baseline]

    # ------------------------------------------------------------------
    # 1 · COLLECT GAPS PER GROUP
    # ------------------------------------------------------------------
    abs_by_group = {}   # {group: {approach: [gaps]}}
    rel_by_group = {}   # same for % gaps

    for ds_name, df in df_dict.items():
        # sub: DataFrame whose *rows are groups* and columns are the approaches
        sub = df["mean"][approaches]

        for group_name, row in sub.iterrows():
            base_val = row[baseline]
            if pd.isnull(base_val):
                continue

            # make sure the inner dicts exist
            abs_by_group.setdefault(group_name, {a: [] for a in comp_approaches})
            rel_by_group.setdefault(group_name, {a: [] for a in comp_approaches})

            for a in comp_approaches:
                val = row[a]
                if pd.notnull(val):
                    gap     = base_val - val                      # +ve ⇒ better than DEF
                    gap_pct = (gap / base_val) * 100
                    abs_by_group[group_name][a].append(gap)
                    rel_by_group[group_name][a].append(gap_pct)

    # ------------------------------------------------------------------
    # 2 · BUILD SUMMARY TABLE  (mean ± SD)
    # ------------------------------------------------------------------
    rows = []
    for group in sorted(abs_by_group):                      # one row per group
        row_dict = {"group": group}
        for a in comp_approaches:
            abs_vals = abs_by_group[group][a]
            rel_vals = rel_by_group[group][a]

            # If the group lacks observations for an approach, fill with NaN
            if abs_vals:
                abs_mean = np.mean(abs_vals)
                abs_sd   = np.std(abs_vals, ddof=1) / len(abs_vals)
                rel_mean = np.mean(rel_vals)
                rel_sd   = np.std(rel_vals, ddof=1) / len(abs_vals)
            else:
                abs_mean = abs_sd = rel_mean = rel_sd = np.nan

            # column names like "AEU_abs_mean", "AEU_abs_sd", ...
            row_dict[f"{a}_abs_m"] = abs_mean
            row_dict[f"{a}_abs_s"]   = abs_sd
            row_dict[f"{a}_rel_m"] = rel_mean
            row_dict[f"{a}_rel_s"]   = rel_sd

        rows.append(row_dict)

    summary_df = pd.DataFrame(rows)

    # nicer column order: group first, then each approach’s stats
    ordered_cols = (
        ["group"] +
        [f"{a}_{suffix}"
         for a in comp_approaches
         for suffix in ("abs_m", "abs_s", "rel_m", "rel_s")]
    )
    summary_df = summary_df[ordered_cols]

    # ------------------------------------------------------------------
    # 3 · DISPLAY OR SAVE
    # ------------------------------------------------------------------
    pd.set_option("display.float_format", "{:,.4f}".format)
    #print(summary_df.to_markdown(tablefmt="github", floatfmt=".2f"))            # to screen


    categories      = list(abs_by_group.keys())[1:]       # 12 categorical groups
    n_groups        = len(categories)
    n_approaches    = len(comp_approaches)

    x_base          = np.arange(n_groups)                # positions on x‑axis
    offset          = 0.8 / n_approaches                  # spread approaches across the group slot
    markers         = ["o", "v", "s"]#, "^", "v", "P", "X"] # enough unique markers
    legend = {
        "TRUE": ["o", "#949494ff"],
        "DEF": ["v", "#949494ff"],
        "DEF-DATA": ["s", "#949494ff"],

        "AEU": ["o", "#4d4da6ff"],
        "AEC": ["v", "#4d4da6ff"],
        "ALU": ["s", "#4d4da6ff"],
        "ALC": ["D", "#4d4da6ff"],

        "CXU": ["o", "#a64da6ff"],
        "CEC": ["v", "#a64da6ff"],
        "CLC": ["s", "#a64da6ff"],

        "ENS": ["s", "#4da6a6ff"],
    }


    plt.figure(figsize=(max(6, n_groups * 1.5), 5))

    for idx, approach in enumerate(comp_approaches):
        means, ses = [], []
        for g in categories:
            if scatter_type == "relative":
                vals = rel_by_group[g][approach]
                x_ticks = np.arange(-10, 30, 5)
            else:
                vals = abs_by_group[g][approach]
                x_ticks = np.arange(-4, 10, 2)
            if vals:
                means.append(np.mean(vals))
                ses.append(np.std(vals, ddof=1) / np.sqrt(len(vals)))
            else:
                means.append(np.nan)
                ses.append(np.nan)

        # x positions shifted per approach
        x_pos = x_base - 0.4 + offset/2 + idx * offset
        plt.errorbar(x_pos, means, yerr=ses, fmt=legend[approach][0], color=legend[approach][1],
                     capsize=3, label=approach, linestyle='None')

    # Aesthetics
    plt.axhline(0, color='grey', linestyle=':')
    plt.xlim(-0.5, n_groups-0.5)
    plt.xticks(x_base, categories, ha='center', fontsize=15)
    plt.yticks(x_ticks, fontsize=15)
    plt.legend(ncol=min(n_approaches, 11), fontsize=15)
    plt.tight_layout()
    plt.savefig(os.path.join(FIGURES_PATH, f"criteria_{scatter_type}_diff_per_approach.pdf"))
    plt.show()
```

# Zero–one Loss Reductions of LFC Approaches per HPS Criterion

For each HPS criterion ($x$-axis), the scatter plot displays the mean and standard error of an approach's reduction in zero–one loss ($y$-axis) compared to majority voting (mv) trained with the default (DEF) HPC 

```python
for scatter_type in ["absolute", "relative"]: 
    # ------------------------------------------------------------------
    # SETTINGS
    # ------------------------------------------------------------------
    approaches      = APPROACHES[1:]
    baseline        = "majority-vote"
    comp_approaches = approaches
    criteria        = list(subsets.keys())
    criteria.insert(1, "DEF-DATA")
    criteria.insert(2, "DEF")
    # print(criteria)

    # ------------------------------------------------------------------
    # 1 · COLLECT GAPS PER GROUP
    # ------------------------------------------------------------------
    abs_by_group = {}   # {group: {approach: [gaps]}}
    rel_by_group = {}   # same for % gaps

    for ds_name, df in df_dict.items():
        # sub: DataFrame whose *rows are groups* and columns are the approaches
        sub = df["mean"][criteria].T[approaches]
        base_val = sub["majority-vote"]["DEF"]

        for group_name, row in sub.iterrows():
            #base_val = row[baseline]
            if pd.isnull(base_val):
                continue

            # make sure the inner dicts exist
            abs_by_group.setdefault(group_name, {a: [] for a in comp_approaches})
            rel_by_group.setdefault(group_name, {a: [] for a in comp_approaches})

            for a in comp_approaches:
                val = row[a]
                if pd.notnull(val):
                    gap     = base_val - val                      # +ve ⇒ better than DEF
                    gap_pct = (gap / base_val) * 100
                    abs_by_group[group_name][a].append(gap)
                    rel_by_group[group_name][a].append(gap_pct)

    # ------------------------------------------------------------------
    # 2 · BUILD SUMMARY TABLE  (mean ± SD)
    # ------------------------------------------------------------------
    rows = []
    for group in sorted(abs_by_group):                      # one row per group
        row_dict = {"group": group}
        for a in comp_approaches:
            abs_vals = abs_by_group[group][a]
            rel_vals = rel_by_group[group][a]

            # If the group lacks observations for an approach, fill with NaN
            if abs_vals:
                abs_mean = np.mean(abs_vals)
                abs_sd   = np.std(abs_vals, ddof=1) / len(abs_vals)
                rel_mean = np.mean(rel_vals)
                rel_sd   = np.std(rel_vals, ddof=1) / len(abs_vals)
            else:
                abs_mean = abs_sd = rel_mean = rel_sd = np.nan

            # column names like "AEU_abs_mean", "AEU_abs_sd", ...
            row_dict[f"{a}_abs_m"] = abs_mean
            row_dict[f"{a}_abs_s"]   = abs_sd
            row_dict[f"{a}_rel_m"] = rel_mean
            row_dict[f"{a}_rel_s"]   = rel_sd

        rows.append(row_dict)

    summary_df = pd.DataFrame(rows)

    # nicer column order: group first, then each approach’s stats
    ordered_cols = (
        ["group"] +
        [f"{a}_{suffix}"
         for a in comp_approaches
         for suffix in ("abs_m", "abs_s", "rel_m", "rel_s")]
    )
    summary_df = summary_df[ordered_cols]

    # ------------------------------------------------------------------
    # 3 · DISPLAY OR SAVE
    # ------------------------------------------------------------------
    pd.set_option("display.float_format", "{:,.4f}".format)
    #print(summary_df.to_markdown(tablefmt="github", floatfmt=".2f"))            # to screen



    # ------------------------------------------------------------------
    # 4 · COLLAPSE mean & sd  →  "$μ_{\pm σ}$"
    # ------------------------------------------------------------------
    def make_cell(mu, sd):
        if np.isnan(mu) or np.isnan(sd):
            return "--"
        if mu > 0:
            return fr"$+{mu:.4f}_{{\pm {sd:.4f}}}$"
        else:
            return fr"${mu:.4f}_{{\pm {sd:.4f}}}$"

    # If they don’t, raise a gentle error.
    if 'abs_by_group' not in globals():
        raise NameError("`abs_by_group` not found – run the gap‑collection cell first.")

    categories      = abs_by_group.keys()        # 12 categorical groups
    n_groups        = len(categories)
    n_approaches    = len(comp_approaches)

    x_base          = np.arange(n_groups)                # positions on x‑axis
    offset          = 0.8 / n_approaches                  # spread approaches across the group slot
    markers         = ["o", "v", "s"]#, "^", "v", "P", "X"] # enough unique markers
    legend = {
        "majority-vote": ["o", "#949494ff"],
        "dawid-skene": ["v", "#949494ff"],

        "crowd-layer": ["o", "#4d4da6ff"],
        "trace-reg": ["v", "#4d4da6ff"],
        "conal": ["s", "#4d4da6ff"],
        "union-net-a": ["D", "#4d4da6ff"],
        "union-net-b": ["X", "#4d4da6ff"],
        "geo-reg-w": ["P", "#4d4da6ff"],
        "geo-reg-f": ["^", "#4d4da6ff"],

        "crowd-ar": ["o", "#a64da6ff"],
        "madl": ["v", "#a64da6ff"],
        "annot-mix": ["s", "#a64da6ff"],
        "coin-net": ["D", "#a64da6ff"],
    }


    plt.figure(figsize=(max(6, n_groups * 1.5), 5))

    for idx, approach in enumerate(comp_approaches):
        means, ses = [], []
        for g in ordered_criteria:
            if scatter_type == "relative":
                vals = rel_by_group[g][approach]
                yticks = np.arange(-5, 40, 5)
            else:
                vals = abs_by_group[g][approach]
                yticks = np.arange(-2, 14, 2)
            if vals:
                means.append(np.mean(vals))
                ses.append(np.std(vals, ddof=1) / np.sqrt(len(vals)))
            else:
                means.append(np.nan)
                ses.append(np.nan)

        # x positions shifted per approach
        x_pos = x_base - 0.4 + offset/2 + idx * offset
        plt.errorbar(x_pos, means, yerr=ses, fmt=legend[approach][0], color=legend[approach][1],
                     capsize=3, label=approach, linestyle='None')

    # Aesthetics
    plt.axhline(0, color='grey', linestyle=':')
    plt.xlim(-0.5, n_groups-0.5)
    plt.xticks(x_base, categories, ha='center', fontsize=15)
    plt.yticks(yticks, fontsize=15)
    plt.legend(ncol=min(n_approaches, 6), fontsize=15)
    plt.tight_layout()
    plt.savefig(os.path.join(FIGURES_PATH, f"approaches_{scatter_type}_diff_per_criterion.pdf"))
    plt.show()
```
